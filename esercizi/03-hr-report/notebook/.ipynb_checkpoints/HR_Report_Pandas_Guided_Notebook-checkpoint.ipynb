
# HR Report con Pandas — Notebook Guidato

In questo notebook costruiamo **passo passo** un report HR usando *Pandas*:
- Carichiamo i 3 CSV (`employees`, `salaries`, `performance`)
- Uniamo e puliamo i dati
- Calcoliamo aggregazioni per `department`/`role`
- Identifichiamo gli outlier retributivi con regola **IQR**
- Creiamo i **ranking** dei top performer (normalizzazione per reparto)
- Esportiamo un **Excel multi-sheet** + un **CSV** con gli outlier

> Ogni sezione include una **spiegazione** e una **cella eseguibile** per vedere il risultato subito sotto.



## Setup iniziale

Importiamo le librerie, settiamo alcune opzioni di visualizzazione e prepariamo le cartelle locali (`data/`, `output/`).  
Se non hai già i CSV, **creiamo dataset di esempio** così puoi eseguire tutto immediatamente.


```python

from __future__ import annotations

from pathlib import Path
import numpy as np
import pandas as pd

# Opzioni di visualizzazione utili
pd.set_option("display.max_rows", 20)
pd.set_option("display.max_columns", 50)
pd.set_option("display.width", 120)

DATA_DIR = Path("data")
OUT_DIR = Path("output")
OUT_DIR.mkdir(exist_ok=True, parents=True)
DATA_DIR.mkdir(exist_ok=True, parents=True)

# --- Dataset di esempio (usati se non hai già i tuoi CSV) ---
SAMPLE_EMP = """employee_id,first_name,last_name,department,role,hire_date
101,Alice,Rossi,Sales,Manager,2019-03-12
102,Bob,Bianchi,Engineering,Developer,2021-07-01
103,Chiara,Verdi,HR,Analyst,2020-11-05
104,Diego,Neri,Engineering,Developer,2018-02-20
"""

SAMPLE_SAL = """employee_id,base_salary,bonus
101,52000,5000
102,45000,2500
103,38000,1500
104,70000,12000
"""

SAMPLE_PERF_2024 = """employee_id,year,rating,goals_met
101,2024,4.5,8
102,2024,3.2,5
103,2024,4.0,7
104,2024,4.8,10
"""

# Scrive i sample solo se i file non esistono già
(emp := DATA_DIR / "employees.csv")
if not emp.exists():
    emp.write_text(SAMPLE_EMP, encoding="utf-8")

(sal := DATA_DIR / "salaries.csv")
if not sal.exists():
    sal.write_text(SAMPLE_SAL, encoding="utf-8")

(perf := DATA_DIR / "performance.csv")
if not perf.exists():
    perf.write_text(SAMPLE_PERF_2024, encoding="utf-8")

DATA_DIR, OUT_DIR
```


## 1) Caricamento dei dati — `load_data`

**Cosa fa:**  
- Legge i tre CSV in `pandas.DataFrame`  
- Converte `hire_date` in `datetime`  
- Se viene passato un `performance_year`, filtra la tabella delle performance su quell'anno

**Perché serve:** uniformiamo i tipi (date) e lavoriamo solo sull'anno di interesse per KPI coerenti.


```python

from typing import Tuple, Optional
from pathlib import Path

def load_data(
    employees_csv: Path,
    salaries_csv: Path,
    performance_csv: Path,
    performance_year: Optional[int] = None,
) -> Tuple[pd.DataFrame, pd.DataFrame, pd.DataFrame]:
    """Carica i 3 CSV in DataFrame pandas e (opzionale) filtra per anno le performance."""
    df_emp = pd.read_csv(employees_csv, parse_dates=["hire_date"])
    df_sal = pd.read_csv(salaries_csv)
    df_perf = pd.read_csv(performance_csv)
    if performance_year is not None:
        df_perf = df_perf[df_perf["year"] == performance_year].copy()
    return df_emp, df_sal, df_perf

# ESECUZIONE: carico i sample creati sopra (anno 2024)
df_emp, df_sal, df_perf = load_data(DATA_DIR / "employees.csv",
                                    DATA_DIR / "salaries.csv",
                                    DATA_DIR / "performance.csv",
                                    performance_year=2024)

display(df_emp.head())
display(df_sal.head())
display(df_perf.head())
```


## 2) Merge dei dataset — `merge_data`

**Cosa fa:**  
- `LEFT JOIN` tra `employees` → `salaries` su `employee_id`  
- `LEFT JOIN` del risultato con `performance`  
- Sostituisce i `NaN` in `bonus` con `0`  
- Crea `total_comp = base_salary + bonus`

**Perché serve:** mettiamo in una **tabella unica** tutte le info rilevanti per successive pulizie, aggregazioni e scoring.


```python

def merge_data(df_emp: pd.DataFrame, df_sal: pd.DataFrame, df_perf: pd.DataFrame) -> pd.DataFrame:
    """Unisce i tre DataFrame e calcola la retribuzione totale."""
    df = df_emp.merge(df_sal, on="employee_id", how="left")
    df = df.merge(df_perf, on="employee_id", how="left")
    df["bonus"] = df["bonus"].fillna(0)
    df["total_comp"] = df["base_salary"] + df["bonus"]
    return df

df = merge_data(df_emp, df_sal, df_perf)
display(df)
```


## 3) Pulizia — `clean_data`

**Cosa fa:**  
- Rimuove duplicati su `employee_id` (tiene il primo)  
- Garantisce che `hire_date` sia `datetime`  
- **Droppa** righe con `base_salary` mancante e con `rating` mancante (campi critici)  

**Perché serve:** evitiamo anomalie strutturali e incompletezze che renderebbero fuorvianti KPI e ranking.


```python

def clean_data(df: pd.DataFrame) -> pd.DataFrame:
    """Pulisce i dati da duplicati e valori mancanti in campi chiave."""
    df = df.drop_duplicates(subset="employee_id", keep="first").copy()
    if not np.issubdtype(df["hire_date"].dtype, np.datetime64):
        df["hire_date"] = pd.to_datetime(df["hire_date"], errors="coerce")
    df = df.dropna(subset=["base_salary", "rating"])
    return df

df_clean = clean_data(df)
display(df_clean)
```


## 4) Aggregazioni per reparto/ruolo — `aggregate_by_dept_role`

**Cosa fa:** calcola, per coppia (`department`, `role`):  
- `emp_count`  
- `base_salary`: `mean`, `median`, `std`  
- `total_comp`: `mean`, `median`, `std`  
- `rating_mean`  

**Perché serve:** ottenendo KPI sintetici per cluster omogenei (reparto/ruolo) confrontiamo meglio comp e performance.


```python

def aggregate_by_dept_role(df: pd.DataFrame) -> pd.DataFrame:
    """Ritorna KPI aggregati per department/role con nomi colonne espliciti."""
    agg = (
        df.groupby(["department", "role"])
        .agg(
            emp_count=("employee_id", "count"),
            base_salary_mean=("base_salary", "mean"),
            base_salary_median=("base_salary", "median"),
            base_salary_std=("base_salary", "std"),
            total_comp_mean=("total_comp", "mean"),
            total_comp_median=("total_comp", "median"),
            total_comp_std=("total_comp", "std"),
            rating_mean=("rating", "mean"),
        )
        .reset_index()
        .sort_values(["department", "role"])
    )
    return agg

agg = aggregate_by_dept_role(df_clean)
display(agg)
```


## 5) Outlier retributivi (IQR) — `detect_outliers_iqr`

**Cosa fa:** per **ogni** `department` calcola Q1/Q3 e l'**IQR** su `total_comp`.  
Definisce *lower* = Q1 - 1.5·IQR e *upper* = Q3 + 1.5·IQR.  
Un record è outlier se `total_comp` è fuori da `[lower, upper]`.  
Aggiunge la colonna booleana `is_comp_outlier`.

**Perché serve:** gli outlier possono distorcere analisi e vanno evidenziati per audit e decisioni HR.


```python

def detect_outliers_iqr(df: pd.DataFrame) -> pd.DataFrame:
    """Aggiunge colonna booleana 'is_comp_outlier' calcolata per reparto su total_comp."""
    df = df.copy()
    q1 = df.groupby("department")["total_comp"].transform(lambda s: s.quantile(0.25))
    q3 = df.groupby("department")["total_comp"].transform(lambda s: s.quantile(0.75))
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    df["is_comp_outlier"] = (df["total_comp"] < lower) | (df["total_comp"] > upper)
    return df

df_flagged = detect_outliers_iqr(df_clean)
display(df_flagged[["employee_id","department","role","base_salary","bonus","total_comp","is_comp_outlier"]])
```


## 6) Ranking dei performer — `_min_max_norm_by_group` e `build_rankings`

**Logica:**  
- Normalizziamo **per reparto** `rating` e `goals_met` con **min-max**:  
  \(x_{norm} = \frac{x - min\_g}{max\_g - min\_g}\)  (gestendo `max==min`)  
- Calcoliamo `perf_score = 0.7 * rating_norm + 0.3 * goals_norm`  
- Estraiamo i **Top 10 per reparto** e i **Top 10 globali**

**Perché serve:** comparare le performance in modo equo tra reparti con scale diverse.


```python

def _min_max_norm_by_group(s: pd.Series, g: pd.Series) -> pd.Series:
    grp_min = s.groupby(g).transform("min")
    grp_max = s.groupby(g).transform("max")
    denom = (grp_max - grp_min).replace(0, np.nan)
    norm = (s - grp_min) / denom
    return norm.fillna(0.0)

def build_rankings(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
    df = df.copy()
    df["rating_norm"] = _min_max_norm_by_group(df["rating"], df["department"])
    df["goals_norm"] = _min_max_norm_by_group(df["goals_met"], df["department"])
    df["perf_score"] = 0.7 * df["rating_norm"] + 0.3 * df["goals_norm"]

    top_by_dept = (
        df.sort_values(["department", "perf_score"], ascending=[True, False])
        .groupby("department", group_keys=False)
        .head(10)
    )
    top_global = df.sort_values("perf_score", ascending=False).head(10)
    return top_by_dept, top_global

top_by_dept, top_global = build_rankings(df_flagged)
print("— Top per reparto —")
display(top_by_dept[["department","employee_id","first_name","last_name","perf_score"]])
print("— Top globali —")
display(top_global[["employee_id","first_name","last_name","perf_score"]])
```


## 7) Esportazione — `export_report`

**Cosa fa:**  
- Crea un **Excel multi-sheet** con: `KPI`, `Aggregati`, `TopByDept`, `TopGlobal`  
- Salva un **CSV** con soli outlier (`is_comp_outlier = True`)  

**Perché serve:** consegniamo un *deliverable* comodo per management e audit.


```python

from pathlib import Path

OUT_XLSX = OUT_DIR / "hr_summary.xlsx"
OUT_OUTLIERS = OUT_DIR / "hr_outliers.csv"

def export_report(
    df: pd.DataFrame,
    agg_role_dept: pd.DataFrame,
    top_by_dept: pd.DataFrame,
    top_global: pd.DataFrame,
    out_xlsx: Path,
    out_outliers_csv: Path,
) -> None:
    # KPI minimi
    kpi_df = pd.DataFrame(
        {
            "n_employees": [df["employee_id"].nunique()],
            "mean_total_comp": [df["total_comp"].mean()],
            "mean_rating": [df["rating"].mean()],
        }
    )
    # Excel multi-sheet
    with pd.ExcelWriter(out_xlsx) as w:
        kpi_df.to_excel(w, sheet_name="KPI", index=False)
        agg_role_dept.to_excel(w, sheet_name="Aggregati", index=False)
        top_by_dept.to_excel(w, sheet_name="TopByDept", index=False)
        top_global.to_excel(w, sheet_name="TopGlobal", index=False)

    # CSV con outlier
    df[df["is_comp_outlier"]].to_csv(out_outliers_csv, index=False)

# ESECUZIONE export
export_report(df_flagged, agg, top_by_dept, top_global, OUT_XLSX, OUT_OUTLIERS)

print("Scritti:")
print(" -", OUT_XLSX.resolve())
print(" -", OUT_OUTLIERS.resolve())

# Mostro un'anteprima del foglio KPI ricaricandolo dall'Excel
kpi_preview = pd.read_excel(OUT_XLSX, sheet_name="KPI")
display(kpi_preview)
```
